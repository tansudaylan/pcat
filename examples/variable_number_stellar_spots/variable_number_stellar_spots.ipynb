# Variable number of rotating starspots in simulated photometry

PCAT samples a catalog of zero to five dark spots on a star with a known 3.2 day rotation period. The script simulates 25.6 days of 0.1 day cadence Poisson photometry with three injected spots, each a periodic Gaussian dip with its own depth, phase epoch, and width. Births, deaths, and jumps draw spot epochs from a matched-filter density with the matching Hastings correction, and eight chains are run. The minimum spot depth exceeds half the deepest injected spot, so one spot cannot be fit as two overlapping halves. All data are simulated.

```python
from pcat.demo import run_example_script

run_example_script('variable_number_stellar_spots/variable_number_stellar_spots.py')
```
