# Variable number of exoplanets in simulated radial-velocity data

PCAT samples catalogs of Keplerian signals in a simulated 6 year radial-velocity (RV) time series of 120 epochs from two spectrographs with different velocity zero points. Three planets with semi-amplitudes of 12, 4, and 2.5 m/s and periods of 5.77, 42.1, and 410 days are injected, with 2 m/s of stellar jitter beyond the 1.5 m/s reported uncertainties. Births and jumps draw periods from a periodogram-weighted density with the matching Hastings correction, and instrument offsets and jitter are marginalized in the likelihood. All data are simulated.
