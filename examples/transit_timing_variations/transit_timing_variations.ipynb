# Transit-timing variations with PCAT
Fit individually simulated mid-transit times with a sinusoidal timing model and inspect the timing posterior.

```python
import numpy as np
from IPython.display import Image, display
from tdpy.verbosity import print
from pcat.demo import run_example_script
from pcat.plotting import EXAMPLES_ROOT

run_example_script('transit_timing_variations/transit_timing_variations.py', '--smoke')
root = EXAMPLES_ROOT / 'transit_timing_variations'
path = root / 'data' / 'ttv_posterior.npz'
print(f'Reading from {path}...')
with np.load(path) as result:
    transit_epoch = result['transit_epoch'].copy()
    observed_times = result['observed_times'].copy()
    posterior = result['chain'].reshape(-1, 4).copy()
print(f'{len(transit_epoch)} simulated transits, {len(posterior)} PCAT posterior samples')
print('Median timing amplitude [minute]:', np.median(posterior[:, 2]) * 1440)
frame = sorted((root / 'visuals').glob('ttv_posterior_swep*.png'))[-1]
print(f'Reading from {frame}...')
display(Image(filename=str(frame)))
```
