# Catalog association

This seeded simulation plots catalog completeness, purity, and false discovery rate as the angular association radius changes.

```python
from pathlib import Path
import importlib.util
from pprint import pprint
from tdpy.verbosity import print as narrate

relative_script = Path('examples/catalog_association_completeness_purity/catalog_association_completeness_purity.py')
root = next(
    candidate for candidate in (Path.cwd(), *Path.cwd().parents)
    if (candidate / relative_script).is_file()
)
script = root / relative_script
narrate(f'Reading from {script}...')
specification = importlib.util.spec_from_file_location('catalog_association_completeness_purity_example', script)
module = importlib.util.module_from_spec(specification)
specification.loader.exec_module(module)
output_path = script.parent / 'visuals' / 'catalog_association_completeness_purity.png'
summary = module.run_example(output_path)
pprint(summary)
```

## Matching tradeoff

A wider angular radius recovers more true counterparts, but chance alignments lower catalog purity. The magnitude constraint limits contamination without changing the positional search.
