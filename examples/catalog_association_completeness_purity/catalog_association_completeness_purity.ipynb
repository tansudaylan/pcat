# Catalog association

This seeded simulation measures catalog completeness and purity as the angular association radius changes.

```python
from pathlib import Path
import importlib.util
from pprint import pprint

candidates = (
    root / 'examples/catalog_association_completeness_purity/catalog_association_completeness_purity.py'
    for root in (Path.cwd(), *Path.cwd().parents)
)
script = next(path for path in candidates if path.is_file())
print(f'Reading from {script}...')
specification = importlib.util.spec_from_file_location('catalog_association_completeness_purity_example', script)
module = importlib.util.module_from_spec(specification)
specification.loader.exec_module(module)
output_path = script.parent / 'visuals' / 'catalog_association_completeness_purity.png'
summary = module.run_example(output_path)
pprint(summary)
```

## Matching tradeoff

A wider angular radius recovers more true counterparts, but chance alignments lower catalog purity. The magnitude constraint limits contamination without changing the positional search.
