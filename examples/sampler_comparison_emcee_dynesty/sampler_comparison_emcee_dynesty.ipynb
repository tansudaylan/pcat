# PCAT, emcee, and dynesty on a simulated sinusoid

The data are simulated: 40 epochs over 30 d of a 5 m/s sinusoid with a 3.7 d period and 2 m/s Gaussian noise. PCAT, emcee (Foreman-Mackey et al. 2013), and dynesty (Speagle 2020) share one likelihood and uniform priors. The figures compare the marginal posteriors, the fraction of samples in the true period mode, and the effective sample size (ESS) per second and per likelihood call.
