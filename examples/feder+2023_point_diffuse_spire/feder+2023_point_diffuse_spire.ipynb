# Feder et al. (2023): point-plus-diffuse SPIRE inference

This notebook showcases the model family developed in Feder et al. (2023), AJ 166, 98, DOI: 10.3847/1538-3881/ace69b. A scaled synthetic field combines a transdimensional point-source population and a cirrus-like diffuse template in PCAT's generic three-channel image engine; the channels are not calibrated SPIRE bands and this repository does not contain the full PCAT-DE implementation used by the paper. The notebook demonstrates the inference structure without claiming to reproduce the paper's simulations or posterior constraints.

```python
from pcat.demo import load_example_posterior, run_example_script
from tdpy.verbosity import print

run_example_script("feder+2023_point_diffuse_spire/feder+2023_point_diffuse_spire.py")
posterior = load_example_posterior(
    'feder+2023_point_diffuse_spire/pcat_runs/feder2023_point_diffuse_spire/data/outp/feder2023_point_diffuse_spire/gdatfinlpost'
)
print(f"Loaded {posterior.listpostcntpmodl.shape[0]} posterior model samples.")
```
