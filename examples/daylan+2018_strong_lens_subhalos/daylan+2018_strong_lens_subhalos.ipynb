# Daylan et al. 2018 strong-lens mock

Run the variable- and fixed-catalog PCAT fits and inspect figures computed from each returned posterior state, alongside the lens-specific pipeline products. The simulated lens and short smoke chains differ from the published realization and are not expected to reproduce its numerical posterior.

```python
from pcat.demo import load_example_posterior, run_example_script
from tdpy.verbosity import print

run_example_script('daylan+2018_strong_lens_subhalos/daylan+2018_strong_lens_subhalos.py', '--smoke', '--fresh')
posterior_catalog = load_example_posterior(
    'daylan+2018_strong_lens_subhalos/daylan2018_catalog/data/outp/daylan2018_catalog/gdatfinlpost'
)
run_example_script('daylan+2018_strong_lens_subhalos/daylan+2018_strong_lens_subhalos.py', '--smoke', '--fresh', '--one-subhalo')
posterior_fixed = load_example_posterior(
    'daylan+2018_strong_lens_subhalos/daylan2018_one_subhalo/data/outp/daylan2018_one_subhalo/gdatfinlpost'
)
print(f"Loaded {posterior_catalog.listpostcntpmodl.shape[0]} variable-catalog and {posterior_fixed.listpostcntpmodl.shape[0]} fixed-catalog model samples.")
```
