# Unbinned Gaussian-mixture inference
Run PCAT on individual simulated event coordinates, then inspect the sampled component centers.

```python
from pathlib import Path
import numpy as np
from IPython.display import Image, display
from tdpy.verbosity import print
from pcat.demo import run_example_script
from pcat.plotting import EXAMPLES_ROOT

run_example_script('unbinned_gaussian_mixture/unbinned_gaussian_mixture.py', '--smoke')
root = EXAMPLES_ROOT / 'unbinned_gaussian_mixture'
path = root / 'data' / 'unbinned_gmm_posterior.npz'
print(f'Reading from {path}...')
with np.load(path) as result:
    events = result['events'].copy()
    centers = result['chain'].reshape(-1, 2, 2).copy()
print(f'{len(events)} individual events, {len(centers)} PCAT posterior samples')
print('Posterior mean centers:', centers.mean(axis=0))
frame = sorted((root / 'visuals').glob('gmix_events_swep*.png'))[-1]
print(f'Reading from {frame}...')
display(Image(filename=str(frame)))
```
