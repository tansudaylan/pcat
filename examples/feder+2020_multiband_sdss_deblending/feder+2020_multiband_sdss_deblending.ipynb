# Feder et al. (2020): multiband SDSS deblending

This notebook showcases the multiband probabilistic cataloging developed by Feder et al. (2020), AJ 159, 163, DOI: 10.3847/1538-3881/ab74cf. The scaled synthetic run uses PCAT's five-band SDSS mode to demonstrate joint crowded-field inference. It is a pipeline analog, not a numerical reproduction of the paper's mock or M2 analyses.

```python
from pcat.demo import load_example_posterior, run_example_script
from tdpy.verbosity import print

run_example_script("feder+2020_multiband_sdss_deblending/feder+2020_multiband_sdss_deblending.py")
posterior = load_example_posterior(
    'feder+2020_multiband_sdss_deblending/pcat_runs/feder2020_multiband_sdss/data/outp/feder2020_multiband_sdss/gdatfinlpost'
)
print(f"Loaded {posterior.listpostcntpmodl.shape[0]} posterior model samples.")
```

```python
from pcat.demo import display_example_visuals, display_posterior_maps

display_posterior_maps(
    posterior,
    "feder+2020_multiband_sdss_deblending",
    "feder2020_posterior",
)
display_example_visuals(
    "feder+2020_multiband_sdss_deblending",
    "pcat_runs/feder2020_multiband_sdss/visuals/init/datacountsen00evt0.png",
    "pcat_runs/feder2020_multiband_sdss/visuals/init/datacountsen04evt0.png",
    "pcat_runs/feder2020_multiband_sdss/visuals/init/cntpresien00evt0.png",
    "pcat_runs/feder2020_multiband_sdss/visuals/post/finl/pdfnlliken*.png",
    "pcat_runs/feder2020_multiband_sdss/visuals/post/finl/histodim/pdfnhistfluxpop0.png",
    "pcat_runs/feder2020_multiband_sdss/visuals/post/finl/histodim/pdfnhistxpospop0.png",
)
```
