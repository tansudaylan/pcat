# Subpixel point-spread-function reconstruction

PCAT fits a piecewise cubic model to an oversampled point-spread function and evaluates its interpolation error across detector pixels.

```python
from pathlib import Path
import importlib.util
from pprint import pprint

candidates = (
    root / 'examples/subpixel_psf_reconstruction/subpixel_psf_reconstruction.py'
    for root in (Path.cwd(), *Path.cwd().parents)
)
script = next(path for path in candidates if path.is_file())
print(f'Reading from {script}...')
specification = importlib.util.spec_from_file_location('subpixel_psf_reconstruction_example', script)
module = importlib.util.module_from_spec(specification)
specification.loader.exec_module(module)
output_path = script.parent / 'visuals' / 'subpixel_psf_reconstruction.png'
summary = module.run_example(output_path)
pprint(summary)
```

## Reconstruction accuracy

The upper panel compares the analytic profile with PCAT's piecewise-cubic model. The lower panel reports interpolation error as a percentage of the profile peak.
