# Butler et al. (2022): SPIRE SZ component separation

Butler et al. (2022), ApJ, DOI: 10.3847/1538-4357/ac6c04, used PCAT-DE to separate CIB point sources, cirrus, and an SZ template in RX J1347.5-1145. This self-contained scaled synthetic analog exercises point sources plus diffuse templates through PCAT's generic three-channel image engine; its channels are not calibrated SPIRE bands. It omits the archival SPIRE, Bolocam, HST, and Chandra products and does not reproduce the published rSZ constraints.

```python
from pcat.demo import load_example_posterior, run_example_script
from tdpy.verbosity import print

run_example_script("butler+2022_spire_sz_component_separation/butler+2022_spire_sz_component_separation.py")
posterior = load_example_posterior(
    'butler+2022_spire_sz_component_separation/pcat_runs/butler2022_spire_sz/data/outp/butler2022_spire_sz/gdatfinlpost'
)
print(f"Loaded {posterior.listpostcntpmodl.shape[0]} posterior model samples.")
```

```python
from pcat.demo import display_example_visuals, display_posterior_maps

display_posterior_maps(
    posterior,
    "butler+2022_spire_sz_component_separation",
    "butler2022_posterior",
)
display_example_visuals(
    "butler+2022_spire_sz_component_separation",
    "pcat_runs/butler2022_spire_sz/visuals/init/datacountsen00evt0.png",
    "pcat_runs/butler2022_spire_sz/visuals/init/cntpmodlen01evt0.png",
    "pcat_runs/butler2022_spire_sz/visuals/init/cntpresien02evt0.png",
    "pcat_runs/butler2022_spire_sz/visuals/post/finl/pdfnlliken*.png",
    "pcat_runs/butler2022_spire_sz/visuals/post/finl/histodim/pdfnhistfluxpop0.png",
    "pcat_runs/butler2022_spire_sz/visuals/post/finl/histodim/pdfnhistxpospop0.png",
)
```
