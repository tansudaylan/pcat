# Synthetic Mejiro time-delay cosmography

This controlled example builds a four-image SIE, shear, and convergence lens at $z_l=0.5$ and $z_s=1.5$ with $H_{0,\rm ref}=70$ km s$^{-1}$ Mpc$^{-1}$ and $\Omega_m=0.3$. Lenstronomy solves the lens equation and calculates the reference delays. The delays and lens model are stored in a Mejiro `LensedSupernova` object.

A seeded observation adds independent 0.1-day Gaussian timing errors to the relative delays for an injected $H_0=67.4$ km s$^{-1}$ Mpc$^{-1}$. PCAT samples a uniform $H_0$ prior from 40 to 100 km s$^{-1}$ Mpc$^{-1}$. The lens potential and $\Omega_m$ stay fixed, so the model scales delays as $\Delta t(H_0)=\Delta t(H_{0,\rm ref})H_{0,\rm ref}/H_0$. This is a synthetic sampler demonstration, not a cosmographic measurement. Lens-model uncertainty and the mass-sheet degeneracy are omitted.
