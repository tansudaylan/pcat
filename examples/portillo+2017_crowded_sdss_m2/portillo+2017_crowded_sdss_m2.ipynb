# Portillo et al. (2017): crowded SDSS M2 field

This notebook runs a scaled synthetic analog of the crowded-field analysis in Portillo et al. (2017), AJ 154, 132, DOI: 10.3847/1538-3881/aa8565. The paper analyzed a single SDSS band; this compact pipeline check uses PCAT's complete five-band SDSS mode because the current restored mode requires all five bands. It demonstrates transdimensional crowded-field cataloging, but does not use the archival M2 image or reproduce the paper's numerical posterior.

```python
from pcat.demo import load_example_posterior, run_example_script
from tdpy.verbosity import print

run_example_script("portillo+2017_crowded_sdss_m2/portillo+2017_crowded_sdss_m2.py")
posterior = load_example_posterior(
    'portillo+2017_crowded_sdss_m2/pcat_runs/portillo2017_sdss_m2/data/outp/portillo2017_sdss_m2/gdatfinlpost'
)
print(f"Loaded {posterior.listpostcntpmodl.shape[0]} posterior model samples.")
```

```python
from pcat.demo import display_example_visuals, display_posterior_maps

display_posterior_maps(
    posterior,
    'portillo+2017_crowded_sdss_m2',
    'portillo2017_posterior',
)
display_example_visuals(
    "portillo+2017_crowded_sdss_m2",
    "pcat_runs/portillo2017_sdss_m2/visuals/init/datacountsen00evt0.png",
    "pcat_runs/portillo2017_sdss_m2/visuals/init/cntpmodlen00evt0.png",
    "pcat_runs/portillo2017_sdss_m2/visuals/init/cntpresien00evt0.png",
    "pcat_runs/portillo2017_sdss_m2/visuals/post/finl/pdfnlliken*.png",
    "pcat_runs/portillo2017_sdss_m2/visuals/post/finl/histodim/pdfnhistfluxpop0.png",
    "pcat_runs/portillo2017_sdss_m2/visuals/post/finl/histodim/pdfnhistxpospop0.png",
)
```
