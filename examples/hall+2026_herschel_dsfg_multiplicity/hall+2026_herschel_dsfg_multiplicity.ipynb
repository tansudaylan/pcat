# Hall et al. (2026): Herschel DSFG multiplicity

Hall et al. (2026), ApJ, DOI: 10.3847/1538-4357/ae1e7a, used three-band Herschel PCAT-DE catalog ensembles to infer dusty star-forming galaxy multiplicity toward ACT-selected sources. This scaled synthetic analog uses a crowded field plus diffuse emission in PCAT's generic three-channel image engine; the channels are not calibrated Herschel bands. It does not use the 71 observed targets or reproduce the published multiplicity distribution.

```python
from pcat.demo import load_example_posterior, run_example_script
from tdpy.verbosity import print

run_example_script("hall+2026_herschel_dsfg_multiplicity/hall+2026_herschel_dsfg_multiplicity.py")
posterior = load_example_posterior(
    'hall+2026_herschel_dsfg_multiplicity/pcat_runs/hall2026_dsfg_multiplicity/data/outp/hall2026_dsfg_multiplicity/gdatfinlpost'
)
print(f"Loaded {posterior.listpostcntpmodl.shape[0]} posterior model samples.")
```
