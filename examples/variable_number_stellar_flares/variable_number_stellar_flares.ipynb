# Variable number of stellar flares in a simulated photometric time series

PCAT samples a variable number of Voigt-profile bursts in a simulated TESS-like 2 minute cadence, 1 day photometric time series. The script injects 3 to 6 fast-rise, exponential-decay (FRED) flares with random peak times, amplitudes, and rise/decay time scales on top of a flat quiescent baseline, draws Poisson counts around the expected count rate, and reuses PCAT's 1D spectral-line machinery (time plays the role of the energy axis) to recover a transdimensional catalog of flare-like bursts. The Voigt profile is symmetric, so it approximates rather than reproduces the injected asymmetric FRED shape; PCAT sometimes needs more than one Voigt component to describe one physical flare.
