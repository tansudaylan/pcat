# PCAT image forward model

The scene is simulated: two compact sources with explicit flux rates, a normalized Gaussian point-spread function, and a uniform background. `pcat.image.forward_model_image` returns the intrinsic scene, PSF kernel, and predicted detector image. This deterministic utility does not sample source parameters or draw Poisson noise.
